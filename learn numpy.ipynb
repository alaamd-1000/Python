```python

years = np.array([1960, 1964, 2004, 2011, 2010, 2008, 2015, 2023, 2010])
magnitudes = np.array([9.5, 9.2, 9.1, 9.1, 8.8, 7.9, 7.8, 7.8, 7.0])
# الأماكن بنفس الترتيب:
places = np.array(["Chile (Valdivia)", "Alaska", "Sumatra", "Japan (Tohoku)",
                   "Chile (Maule)", "China (Sichuan)", "Nepal", "Turkey-Syria", "Haiti"])
```

```python
np.random.seed(42)

# 200 زلزال: العمق (كم) والمقدار وعدد الضحايا
depth = np.random.randint(5, 300, size=200)
magnitude = np.round(np.random.uniform(4.0, 8.5, size=200), 1)
casualties = np.random.poisson(lam=magnitude**2, size=200)

# مصفوفة 2D: كل صف زلزال وكل عمود خاصية
quakes = np.column_stack([depth, magnitude, casualties])
print(quakes.shape)   # (200, 3)
```

محاكاة الزلازل 

```python
np.random.seed(42)

depth = np.random.randint(5, 300, size=200)
magnitude = np.round(np.random.uniform(4.0, 8.5, size=200), 1)
casualties = np.random.poisson(lam=magnitude**2, size=200)

quakes = np.column_stack([depth, magnitude, casualties])
```

```python
print("متوسط كل الزلازل:", casualties.mean())

shallow = depth < 70
print("متوسط الزلازل السطحية:", casualties[shallow].mean())
```

```python
order = np.argsort(casualties)[::-1]   # ترتيب تنازلي للمواضع
top10 = order[:10]

print(quakes[top10])
```
